```python
# 1. تثبيت المكتبة الرسمية
!pip install google-genai -q

import sys
from google import genai
from google.genai import types
from google.colab import userdata

# 2. إعداد مفتاح الـ API بشكل آمن من قائمة الأسرار (Secrets)
try:
  API_KEY = userdata.get('GEMINI_API_KEY')
  client = genai.Client(api_key=API_KEY)
except Exception as e:
  print("❌ خطأ: لم يتم العثور على المفتاح السري GEMINI_API_KEY.")
  print("👉 يرجى إضافته من القائمة الجانبية (🔑 Secrets) وتفعيل خيار Notebook access.")
  sys.exit(1)

# 3. قراءة ملف السياسات
try:
  with open("company_policy.txt", "r", encoding="utf-8") as f:
    policy_text = f.read()
except FileNotFoundError:
  print("❌ خطأ: لم يتم العثور على ملف company_policy.txt في بيئة الكولاب.")
  sys.exit(1)

# 4. صياغة التعليمات الصارمة المعدلة (System Instructions)
sys_instructions = f"""
You are an official Company Policy Assistant for Ejmam Transport Co.
Your primary duty is to answer user questions based STRICTLY and ONLY on the provided Company Policy text below.

RULES:
1. Do not use any outside knowledge or assumptions.
2. ALWAYS match the user's language:
   - If asked in Arabic, reply in Arabic.
   - If asked in English, reply in English.
3. If the information is NOT in the policy:
   - Arabic reply: "عذراً، هذه المعلومة غير مذكورة في سياسة الشركة الرسمية."
   - English reply: "I'm sorry, but this information is not mentioned in the official company policy."
4. DO NOT use markdown bold formatting (asterisks like **) in any response. Plain text only.

COMPANY POLICY TEXT:
{policy_text}
"""

# 5. بدء الشات التفاعلي باستخدام النموذج المعتمد
chat = client.chats.create(
    model="gemini-3.6-flash",
    config=types.GenerateContentConfig(
        system_instruction=sys_instructions, temperature=0
    ),
)

print("--- Ejmam Transport Co. Policy Assistant Chatbot ---")
print("Type your question below. Type 'exit' or 'quit' to stop.\n")

# 6. الحلقة التفاعلية المباشرة
while True:
  user_input = input("You: ")
  if user_input.strip().lower() in ["exit", "quit"]:
    print("Bot: Session ended. Have a great day!")
    break
  if not user_input.strip():
    continue

  response = chat.send_message(user_input)

  # تنظيف النص من أي نجوم متبقية لضمان شكل الرد العربي
  clean_reply = response.text.replace("*", "").strip()

  print(f"Bot: {clean_reply}\n")
```
